```python
#deploys pattern recognition to identify and quantify relationship between different yet related items
#used for product recommendations to generate business value

#na = number of times A appeared, nb = number of times B appeared, 
#nab = number of times A and B appeared together, n = total number of observations
#support(A) = na/n
#support(A->B) = nab/na
#confidence(A -> B) = support(A->B)/support(A) #how likely B is purchased given that A was purchased
#lift(A->B) = confidence(A->B)/support(B)
#lift score > 1: if A is purchased, it is likely that B will also be purchased
#lift score < 1: if A is purchased, it is unlikely that B will also be purchased
#lift sore = 1: there is no association between A and B
```

```python
#association rule mining using apriori algorithm
```

```python
import pandas as pd
from mlxtend.frequent_patterns import apriori
from mlxtend.frequent_patterns import association_rules
```

```python
data = pd.read_csv(r"C:\Users\despi\OneDrive\Desktop\Python for DS and ML Part 2\data\groceries.csv")
```

```python
data.head()
```

```python
basket_sets = data.stack().str.get_dummies().groupby(level=0).sum()
```

```python
basket_sets.head()
```

```python
basket_sets_bool = basket_sets.astype(bool)
apriori(basket_sets_bool, min_support = 0.02)
```

```python
basket_sets_bool = basket_sets.astype(bool)
apriori(basket_sets_bool, min_support = 0.02, use_colnames=True)
```

```python
df_combo = basket_sets.astype(bool)

frequent_itemsets = apriori(df_combo, min_support = 0.002, use_colnames=True)

frequent_itemsets['length'] = frequent_itemsets['itemsets'].apply(lambda x: len(x))

frequent_itemsets
```

```python
frequent_itemsets[(frequent_itemsets['length'] >= 3)]
```

```python
rules = association_rules(frequent_itemsets, metric='confidence', min_threshold = 0.9)
rules.head()
```

```python
rules = association_rules(frequent_itemsets, metric='lift', min_threshold = 1)
rules.head()
```

```python
rules[(rules['lift']>=5) & (rules['confidence']>=0.8)]
```
