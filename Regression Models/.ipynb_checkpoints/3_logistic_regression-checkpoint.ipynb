```python
titanic_training = pd.read_csv(address)
titanic_training.columns = ['PassengerId', 'Survived', 'Pclass','Name','Sex', 'Age', 'SibSp', 'Parch','Ticket','Fare','Cabin','Embarked']
```

```python
#is our target variable binary?
sns.countplot(x='Survived', hue='Survived', data=titanic_training, palette='hls', legend=False)
```

```python
#are there any missing values?
titanic_training.isnull().sum()
```

```python
titanic_training.describe()
```

```python
titanic_data=titanic_training.drop(['Name','Ticket','Cabin'],axis=1)
titanic_data.head()
```

```python
#missing values inputation
sns.boxplot(x='Parch',y='Age', data=titanic_data)
```

```python
Parch_groups = titanic_data.groupby(titanic_data['Parch'])
Parch_groups.mean(numeric_only = True)
```

```python
#inpute age depending on the Parch values
def age_approx(cols):
    Age = cols.iloc[0]
    Parch = cols.iloc[1]

    if pd.isnull(Age):
        if Parch == 0:
            return 32
        elif Parch == 1:
            return 34
        elif Parch == 2:
            return 17
        elif Parch ==3:
            return 33
        elif Parch == 4:
            return 45
        else:
            return 30

    else:
        return Age
        
```

```python
titanic_data['Age'] = titanic_data[['Age','Parch']].apply(age_approx, axis=1)
titanic_data.isnull().sum()
```

```python
titanic_data.dropna(inplace=True)
titanic_data.reset_index(inplace=True,drop=True)

print(titanic_data.info())
```

```python
#re-encode variables
#AGe and Embark need to be re-encoded to numerical categorical variables

from sklearn.preprocessing import LabelEncoder
label_encoder = LabelEncoder()
gender_cat = titanic_data['Sex']
gender_encoded = label_encoder.fit_transform(gender_cat)
gender_encoded[0:5]
```

```python
titanic_data.head()
```

```python
#so 1 = male, 0 = female
```

```python
gender_df = pd.DataFrame(gender_encoded, columns = ['male_gender'])
gender_df.head()
```

```python
#encode Embarked variable
embarked_cat = titanic_data['Embarked']
embarked_encoded = label_encoder.fit_transform(embarked_cat)
embarked_encoded[:100]
```

```python
#so 2= S, 0 =C, 1 = Q
```

```python
#this is a multinomial categorical variable
#we need to get binary output here
#we'll use one-hot encoding to have 3 binomial columns
```

```python
from sklearn.preprocessing import OneHotEncoder
binary_encoder = OneHotEncoder(categories = 'auto')
embarked_1hot = binary_encoder.fit_transform(embarked_encoded.reshape(-1,1))
embarked_1hot_mat = embarked_1hot.toarray()
embarked_df = pd.DataFrame(embarked_1hot_mat, columns = ['C','Q','S'])
embarked_df.head()
```

```python
titanic_data.drop(['Sex','Embarked'], axis = 1, inplace=True)
titanic_data.head()
```

```python
titanic_dmy = pd.concat([titanic_data, gender_df, embarked_df], axis = 1, verify_integrity=True).astype(float)
titanic_dmy[:5]
```

```python
#check for independence between features
sns.heatmap(titanic_dmy.corr())
```

```python
titanic_dmy.drop(['Fare'], axis=1,inplace=True)
titanic_dmy.head()
```

```python
#need at least 50 observations per feature 
#check that
titanic_dmy.info()
```

```python
#model deployment
X_train, X_test, y_train, y_test = train_test_split(titanic_dmy.drop('Survived', axis=1),
                                                    titanic_dmy['Survived'], test_size = 0.20,
                                                    random_state = 200)
```

```python
print(X_train.shape)
print(y_train.shape)
```

```python
X_train.head(5)
```

```python
LogReg=LogisticRegression(solver='liblinear')
LogReg.fit(X_train.values, y_train.values)
```

```python
#use the model to make predictions
y_pred = LogReg.predict(X_test.values)
```

```python
#model evaluation
#use classification report and confusion matrices
print(classification_report(y_test,y_pred))
```

```python
#confusion matrix
y_train_pred = cross_val_predict(LogReg, X_train, y_train, cv=5)
confusion_matrix(y_train, y_train_pred)
```

```python
precision_score(y_train, y_train_pred)
```
